```python
import pandas as pd

df = pd.read_csv(r"C:\Users\user\Downloads\synthetic_dataset.csv")

df['Rating'] = df['Rating'].fillna(df['Rating'].median())

print(df)
```

```python
df['Stock'] = df['Stock'].ffill()
print("\nAfter forward fill:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Stock_Encoder'] = encoder.fit_transform(df['Stock'])
print(df)
print(encoder.classes_)
```

```python
encoded_df = pd.get_dummies(df, columns=['Stock'])
print(encoded_df)
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=['Stock'],
    dtype=int
)
print(encoded_df)    
```
